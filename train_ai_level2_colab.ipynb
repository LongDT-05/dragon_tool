# 🐉 Huấn Luyện AI Cấp 2 (YOLOv8-nano) cho DaoRong_tool v12.0 (100% Miễn Phí trên Google Colab)

**Quy trình 3 bước siêu gọn:**
1. Bật công tắc **Thu thập mẫu ảnh tự động (`ai_dataset/`)** trong tab *Cài Đặt* của DaoRong_tool cho đến khi có ~150–300 ảnh.
2. Gán nhãn miễn phí bằng [Roboflow](https://roboflow.com/) hoặc LabelImg với 8 nhãn chuẩn (`egg`, `altar`, `popup_close_x`, `worldmap_home`, `breed_button`, `bag_full_popup`, `disconnect_popup`, `parent_dragon`), rồi xuất bộ dữ liệu định dạng **YOLOv8 ZIP**.
3. Tải file `dataset.zip` lên Google Colab này (chọn **Runtime -> Change runtime type -> T4 GPU**) và bấm **Run all**. Sau ~5 phút, tải file `daorong_yolo_v2.onnx` về bỏ vào thư mục `models/` cạnh `DaoRong_tool.exe`!

```python
!pip install -q ultralytics onnx onnxslim
import os, zipfile
from google.colab import files

print('Hãy chọn file dataset.zip (định dạng YOLOv8 có data.yaml):')
uploaded = files.upload()
zip_name = next(iter(uploaded.keys()))
with zipfile.ZipFile(zip_name, 'r') as z:
    z.extractall('/content/dataset')
print('Đã giải nén xong vào /content/dataset!')
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
model.train(
    data='/content/dataset/data.yaml',
    epochs=80,
    imgsz=640,
    batch=16,
    name='daorong_v2'
)
best = YOLO('runs/detect/daorong_v2/weights/best.pt')
onnx_path = best.export(format='onnx', imgsz=640, simplify=True, opset=12)
print('Xuất ONNX thành công:', onnx_path)
```

```python
import shutil
from google.colab import files

shutil.copy('runs/detect/daorong_v2/weights/best.onnx', '/content/daorong_yolo_v2.onnx')
files.download('/content/daorong_yolo_v2.onnx')
print('Hãy chép daorong_yolo_v2.onnx vào thư mục models/ cạnh DaoRong_tool.exe để kích hoạt AI Cấp 2!')
```
